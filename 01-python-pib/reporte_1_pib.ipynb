Nombre o descripción de la variable: Producto Interno Bruto (PIB) nominal a precios corrientes

Una breve explicación de qué representa la serie: El PIB representa el valor de todos los bienes y servicios finales producidos en una economía en un tiempo determinado, en este caso la economía chilena. Como está en precios corrientes significa que se calcula con los precios vigentes, o sea, que no estan ajustados a la inflación. Los valores se miden en miles de millones de pesos. Como es empalmada significa que se ajustó para poder comparar los datos a lo largo del tiempo sin interrupciones. La fuente es el Banco Central de Chile.

```python
#2. Importación de bibliotecas
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#3. Carga del archivo
PIB = "PIB(in).csv"
df = pd.read_csv(PIB)
```

```python
#4. Exploración de los datos
print("Primeras 5 observaciones")
display(df.head())

print("Nombre de las columnas")
print(df.columns.tolist())

print(f"Cantidad de filas y columnas:{df.shape}")
print("Tipos de datos")
print(df.dtypes)

print("Valores faltantes por columna")
print(df.isnull().sum())
```

```python
#5. Preparación de la serie
columna_fecha = "Fecha"
df[columna_fecha] = pd.to_datetime(df[columna_fecha])
df = df.sort_values(by=columna_fecha)

columna_valor = "Valor"
```

```python
#6. Elaboración del gráfico
plt.figure(figsize=(12, 6))
plt.plot(df[columna_fecha], df[columna_valor], label= "Evolución del PIB", color="purple")
plt.xticks(df[columna_fecha][::8], rotation=45)
plt.title("Evolución del PIB nominal trimestral de Chile (1996–2026)", fontsize=14)
plt.xlabel("Fecha", fontsize=12)
plt.ylabel("Valor en miles de millones de pesos", fontsize=12)
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.figtext(0.99, 0.01, "Fuente: Banco Central de Chile", horizontalalignment = "right", fontsize=10)
plt.tight_layout()
plt.show()
```

¿Cuál es la tendencia general de la serie? La tendencia general de la serie es a aumentar conforme pasa el tiempo

¿Se observan aumentos o disminuciones importantes? 
Se puede observar una disminución importante durante el año 2020, el cual es atribuible a la pandemia, luego de esto se ve que el PIB comienza a aumentar de manera acelerada

¿Existen períodos con valores especialmente altos o bajos?
Podemos ver que los valores más bajos se registran al inicio de la serie, siendo menos de 1/4 del valor que tiene actualmente. Los valores más altos se encuentran al final del período. Cabe destacar que al ser precios corrientes, estos aumentos no son solo por actividad económica, sino que también se debe al efecto acumulado de la inflación a lo largo de los años.

Se puede apreciar una forma de serrucho que se repite año tras año, esto es dada la estacionalidad trimestral que presenta la economía chilena, nos indica que hay momentos del año en los cuales la producción de bienes y servicios es mayor
